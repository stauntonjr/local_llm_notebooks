Tested with an RTX Pro 6000.

# vLLM

```python
%%bash
uv pip install vllm
```

## MTP

```python
%%bash
vllm serve Qwen/Qwen3.6-27B --host 127.0.0.1 --max-model-len 262144 --reasoning-parser qwen3 --speculative-config '{"method":"qwen3_next_mtp","num_speculative_tokens":5}'
```

# llama.cpp

## Installation

```python
%%bash
git clone https://github.com/ggml-org/llama.cpp
cmake llama.cpp -B llama.cpp/build \
    -DBUILD_SHARED_LIBS=OFF -DGGML_CUDA=ON
cp llama.cpp/build/bin/llama-* llama.cpp
```

## MTP

```python
%%bash
./llama.cpp/llama-server \
    -hf unsloth/Qwen3.6-27B-MTP-GGUF:UD-Q4_K_XL \
    -ngl 99 -c 262000 -fa on -np 1 \
    --spec-type draft-mtp --spec-draft-n-max 5
```
